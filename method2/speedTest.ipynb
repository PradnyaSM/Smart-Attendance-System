```python
import time
start_time = time.time()
```

```python
import cv2
print("time==",time.time()-start_time)
cap = cv2.VideoCapture(0)# 0 for primary camera,1 for secondary devices 
print("initialize time==",time.time()-start_time)
while cap.isOpened():

    ret, frame = cap.read()
    print("record time==",time.time()-start_time)
    if not ret:  # Check if the frame was successfully read
        print("End of video or cannot read the frame.")
        break
    
    frame = cv2.resize(frame,(1200,700))

    cv2.imshow("frame",frame)
    print("show time==",time.time()-start_time)
    
    k = cv2.waitKey(1) & 0xFF # waitkey==0 -->static img, ==1--> dynamic vdo
    if (k==ord('s') ):
        break
cap.release()
cv2.destroyAllWindows()
```
